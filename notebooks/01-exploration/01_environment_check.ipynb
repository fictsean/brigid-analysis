# Environment Check

Verify all key packages are importable and at expected versions.

```python
import sys
import importlib.metadata

print(f"Python: {sys.version}")

import numpy as np
import pandas as pd
import xarray as xr
import geopandas as gpd
import matplotlib
import scipy
import sklearn
import statsmodels
import intake_esm
import zarr

def pkg_version(name):
    try:
        return importlib.metadata.version(name)
    except importlib.metadata.PackageNotFoundError:
        return "not found"

packages = [
    ("numpy",        np.__version__),
    ("pandas",       pd.__version__),
    ("xarray",       xr.__version__),
    ("geopandas",    gpd.__version__),
    ("matplotlib",   matplotlib.__version__),
    ("scipy",        scipy.__version__),
    ("scikit-learn", sklearn.__version__),
    ("statsmodels",  statsmodels.__version__),
    ("cdsapi",       pkg_version("cdsapi")),
    ("intake-esm",   intake_esm.__version__),
    ("zarr",         zarr.__version__),
]

print()
for pkg, ver in packages:
    print(f"  {pkg:<20} {ver}")
```

```python
# Quick smoke test: xarray + numpy round-trip
import numpy as np
import xarray as xr

da = xr.DataArray(
    np.random.randn(5, 10),
    dims=["lat", "lon"],
    attrs={"units": "K", "long_name": "Surface temperature anomaly"},
)
print(da)
print(f"\nMean: {da.mean().item():.4f} K")
```
